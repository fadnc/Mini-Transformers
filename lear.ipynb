```python
import numpy as np
```

```python
tokens = ["the", "cat", "sleeps"]

X = np.array([
    [1.0, 0.0, 0.0],  # the
    [0.0, 1.0, 0.0],  # cat
    [0.0, 0.8, 0.6]   # sleeps
])

print(X.shape)
```

```python
def dot_prod(a, b):
    result = 0
    
    for i in range(len(a)):
        result += a[i] * b[i]
    
    return result

cat = X[1]
sleeps = X[2]

print(dot_prod(cat, sleeps))
```

```python
scores = X @ X.T

print(scores)
```

```python
def softmax(x):
    x = x - np.max(x, axis=-1, keepdims=True)  # subrtracting max val to prevent numerical overflow
    
    exp_x = np.exp(x)
    
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)

weights = softmax(scores)

print(np.round(weights, 3))
```

```python
output = weights @ X

print(np.round(output, 3))
```
